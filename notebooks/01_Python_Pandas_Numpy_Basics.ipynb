## 💡 1. Kenapa Pandas & NumPy?
Bayangkan **Pandas** seperti program Microsoft Excel yang super cepat dan bisa mengolah jutaan baris data hanya dengan 1 baris kode. Sedangkan **NumPy** adalah mesin kalkulator super cepat di belakang layar yang menangani perhitungan angka dan matriks.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Library berhasil diimpor!")
```

## 📊 2. Membuat & Memanipulasi DataFrame Sederhana
Mari kita buat tabel simulasi transaksi toko elektronik sederhana:

```python
# Membuat data contoh menggunakan dictionary
data_toko = {
    'Nama_Pelanggan': ['Andi', 'Budi', 'Citra', 'Dewi', 'Eko', 'Fani'],
    'Kota': ['Jakarta', 'Surabaya', 'Bandung', 'Jakarta', 'Surabaya', 'Medan'],
    'Usia': [25, 34, 29, 42, 19, 31],
    'Total_Belanja_Juta': [1.5, 4.2, 0.8, 7.5, 1.2, 3.4],
    'Metode_Bayar': ['QRIS', 'Transfer', 'Tunai', 'Kartu Kredit', 'QRIS', 'Transfer']
}

df = pd.DataFrame(data_toko)
df
```

## 🔍 3. Memfilter & Menganalisis Data
Misalnya kita ingin melihat pelanggan dari kota Jakarta atau pelanggan yang usianya di atas 30 tahun:

```python
# Filter pelanggan dengan usia > 30 tahun
pelanggan_senior = df[df['Usia'] > 30]
print("Pelanggan di atas 30 tahun:")
pelanggan_senior
```

```python
# Rata-rata belanja per Kota
rata_belanja_kota = df.groupby('Kota')['Total_Belanja_Juta'].mean().reset_index()
rata_belanja_kota
```

## 📈 4. Visualisasi Data Cepat dengan Seaborn

```python
plt.figure(figsize=(7, 4))
sns.barplot(data=rata_belanja_kota, x='Kota', y='Total_Belanja_Juta', palette='viridis')
plt.title('Rata-rata Total Belanja Berdasarkan Kota (Juta Rp)')
plt.ylabel('Belanja (Juta)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
