## 🧱 2. Python Esensial: Variabel, Struktur Data, Perulangan & Fungsi

### A. Variabel & 4 Tipe Data Primitif
Bayangkan **Variabel** seperti kardus penyimpanan yang Anda beri label nama:
- **String (Teks)** : Harus diapit tanda petik (`"Jakarta"`)
- **Integer (Bilangan Bulat)** : Angka tanpa koma (`28`)
- **Float (Bilangan Desimal)** : Angka pecahan dengan titik (`7.5`)
- **Boolean (Logika)** : Hanya bernilai Benar (`True`) atau Salah (`False`)

```python
# Menyimpan data pelanggan ke variabel sederhana
nama_pelanggan = "Budi Santoso"      # String (Teks)
usia = 29                           # Integer (Angka Bulat)
total_belanja_juta = 4.75           # Float (Angka Desimal)
member_aktif = True                 # Boolean (True/False)

print(f"Pelanggan : {nama_pelanggan}")
print(f"Usia      : {usia} tahun (Tipe: {type(usia).__name__})")
print(f"Belanja   : Rp {total_belanja_juta} Juta (Tipe: {type(total_belanja_juta).__name__})")
print(f"Member    : {member_aktif} (Tipe: {type(member_aktif).__name__})")
```

### B. Wadah Data Koleksi: List, Tuple & Dictionary

1. **List `[ ... ]`**: Wadah daftar belanja berurut yang bisa diubah isinya. Indeksnya selalu mulai dari **0**!
2. **Slicing List `[awal : akhir]`**: Memotong bagian list (batas akhir tidak ikut diambil).
3. **Tuple `( ... )`**: Mirip List, tetapi isinya terkunci permanen tidak bisa diubah (*immutable*).
4. **Dictionary `{ kunci : nilai }`**: Wadah berpasangan seperti buku kontak telepon.

```python
# 1. List dan Slicing (Pemotongan Indeks)
kota_cabang = ['Jakarta', 'Surabaya', 'Bandung', 'Medan', 'Semarang']
print(f"Kota Pertama (indeks 0) : {kota_cabang[0]}")
print(f"Kota Terakhir (indeks -1): {kota_cabang[-1]}")
print(f"Slicing 3 Kota Pertama   : {kota_cabang[0:3]}")  # Indeks 0, 1, 2

# 2. Tuple (Permanen)
koordinat_toko = (-6.2088, 106.8456) # Latitude & Longitude Jakarta
print(f"Koordinat Toko (Tuple)   : {koordinat_toko}")

# 3. Dictionary (Profil Nasabah)
profil_nasabah = {
    'nama': 'Citra Lestari',
    'pekerjaan': 'Wiraswasta',
    'penghasilan_juta': 12.5,
    'skor_kredit': 750
}
print(f"Nama Nasabah             : {profil_nasabah['nama']}")
print(f"Skor Kredit              : {profil_nasabah['skor_kredit']}")
```

### C. Perulangan Singkat (`for loop`) & List Comprehension
- **`for loop`**: Melakukan pekerjaan berulang pada setiap elemen list.
- **List Comprehension**: Cara khas Python menulis loop hanya dalam 1 baris elegan!

```python
# Contoh: Mengubah daftar belanja dari Dollar ($) ke Rupiah (Rp 15.000 / $)
belanja_usd = [10, 25, 50, 100]

# 1. Menggunakan For Loop Biasa:
belanja_idr_loop = []
for b in belanja_usd:
    belanja_idr_loop.append(b * 15000)
print(f"Hasil For Loop Biasa       : {belanja_idr_loop}")

# 2. Menggunakan List Comprehension (1 Baris Ringkas!):
belanja_idr_ringkas = [b * 15000 for b in belanja_usd]
print(f"Hasil List Comprehension   : {belanja_idr_ringkas}")
```

### D. Logika Percabangan (`if-elif-else`) & Fungsi Sederhana (`def`)

- **Percabangan (`if-else`)**: Memberikan instruksi berbeda berdasarkan kondisi (seperti kasir bioskop memeriksa usia penonton).
- **Fungsi (`def`)**: Resep mesin yang bisa dipanggil berulang kali tanpa mengetik ulang rumusnya.

```python
# Membuat fungsi kalkulator kategori risiko kredit nasabah
def cek_status_pinjaman(penghasilan, skor_kredit):
    if penghasilan >= 10.0 and skor_kredit >= 700:
        return "Disetujui Otomatis (Nasabah Prioritas) 🟢"
    elif penghasilan >= 5.0 and skor_kredit >= 600:
        return "Butuh Peninjauan Manual 🟡"
    else:
        return "Ditolak (Risiko Tinggi) 🔴"

# Uji coba fungsi buatan kita:
status_1 = cek_status_pinjaman(12.5, 750)
status_2 = cek_status_pinjaman(4.0, 550)

print(f"Nasabah 1: {status_1}")
print(f"Nasabah 2: {status_2}")
```

## ⚡ 3. Mengenal NumPy: Mesin Kalkulator Angka Super Cepat

### Mengapa Praktisi Data Wajib Pakai NumPy?
Jika Anda memiliki 10.000 data transaksi dan ingin memberi diskon 10%, menggunakan perulangan `for loop` Python biasa membutuhkan waktu lama.  
Dengan **NumPy Array**, operasi matematika dihitung serentak dalam sekejap mata (*Vectorized Operations*)!

```python
import numpy as np

# Membuat array harga belanjaan dalam ribuan rupiah
harga = np.array([50, 100, 150, 200, 250, 300])

# Operasi Vektor: Diskon 10% serentak untuk semua angka
harga_diskon = harga * 0.9

print(f"Harga Asli   : {harga}")
print(f"Harga Diskon : {harga_diskon}")

# Fungsi Statistik NumPy yang Sering Dipakai di Data Mining:
print("\n--- Rangkuman Statistik NumPy ---")
print(f"Rata-rata (Mean)         : Rp {np.mean(harga):.1f} Ribu")
print(f"Nilai Tengah (Median)    : Rp {np.median(harga):.1f} Ribu")
print(f"Nilai Minimum (Min)      : Rp {np.min(harga)} Ribu")
print(f"Nilai Maksimum (Max)     : Rp {np.max(harga)} Ribu")
print(f"Standar Deviasi (Std)    : {np.std(harga):.2f}")
print(f"Persentil ke-75          : Rp {np.percentile(harga, 75)} Ribu")
```

## 🐼 4. Jantung Data Mining: Membedah Pandas DataFrame

Bayangkan **Pandas** seperti program Microsoft Excel yang super bertenaga tanpa batasan baris:
- **Series**: Satu kolom data tunggal (1 Dimensi).
- **DataFrame**: Tabel lengkap baris dan banyak kolom (2 Dimensi).

Mari kita buat tabel transaksi toko ritel:

```python
import pandas as pd

# Membuat DataFrame dari Dictionary gabungan List
data_toko = {
    'ID_Transaksi': [101, 102, 103, 104, 105, 106, 107, 108],
    'Bulan': ['Jan', 'Jan', 'Feb', 'Feb', 'Mar', 'Mar', 'Apr', 'Apr'],
    'Nama_Pelanggan': ['Andi', 'Budi', 'Citra', 'Dewi', 'Eko', 'Fani', 'Gita', 'Hadi'],
    'Kota': ['Jakarta', 'Surabaya', 'Bandung', 'Jakarta', 'Surabaya', 'Medan', 'Bandung', 'Jakarta'],
    'Usia': [25, 34, 29, 42, 19, 31, 23, 38],
    'Kategori_Produk': ['Laptop', 'HP', 'Aksesoris', 'Laptop', 'HP', 'Aksesoris', 'HP', 'Laptop'],
    'Total_Belanja_Juta': [15.0, 7.5, 0.8, 18.0, 6.5, 1.2, 8.0, 22.0],
    'Metode_Bayar': ['QRIS', 'Transfer', 'Tunai', 'Kartu Kredit', 'QRIS', 'Transfer', 'QRIS', 'Kartu Kredit'],
    'Status_Member': ['Silver', 'Gold', 'Silver', 'Platinum', 'Silver', 'Gold', 'Silver', 'Platinum']
}

df = pd.DataFrame(data_toko)
print("=== TABEL DATA TRANSAKSI TOKO ELEKTRONIK RITEL ===")
df
```

### 💡 Kiat Nyata: Cara Membaca File Excel & CSV dari Komputer

Di dunia nyata, Anda tidak mengetik data manual, melainkan membaca file:
- Membaca file CSV biasa : `df = pd.read_csv('nama_file.csv')`
- **Jebakan CSV Indonesia/Eropa**: Di Indonesia, file Excel sering diekspor dengan pemisah **titik koma (`;`)**, bukan koma.  
  👉 Solusinya: tambahkan parameter `sep=';'`, contoh: `df = pd.read_csv('data.csv', sep=';')`
- Membaca file Excel : `df = pd.read_excel('nama_file.xlsx')`
- Membaca file JSON : `df = pd.read_json('nama_file.json')`

### 🔍 6 Perintah Sakti Inspeksi Karakteristik Data:

```python
# 1. Dimensi data: (Jumlah Baris, Jumlah Kolom)
print(f"Ukuran Tabel (Baris, Kolom): {df.shape}")

# 2. Daftar nama kolom
print(f"Kolom Tabel: {df.columns.tolist()}\n")

# 3. Cek Nilai Kosong (Missing Values)
print("--- Pengecekan Data Kosong (.isna().sum()) ---")
print(df.isna().sum())

# 4. .info() : Mengetahui tipe data ringkas
print("\n--- Informasi Tipe Data (.info()) ---")
df.info()
```

```python
# 5. .describe() : Ringkasan statistik otomatis untuk seluruh kolom angka
print("--- Ringkasan Statistik Otomatis (.describe()) ---")
df.describe()
```

```python
# 6. .value_counts() : Menghitung frekuensi sebaran kategori (Sangat Sering Dipakai!)
print("--- Frekuensi Kategori Produk yang Terjual ---")
print(df['Kategori_Produk'].value_counts())

print("\n--- Frekuensi Metode Bayar yang Digunakan ---")
print(df['Metode_Bayar'].value_counts())
```

## 🎯 5. Seni Memilih & Menyaring Data (Slicing, Filtering, .loc vs .iloc)

### A. Memilih Kolom: Tunggal vs Ganda

```python
# 1 Kolom saja -> Menghasilkan Series
kolom_usia = df['Usia']

# Beberapa Kolom -> Menghasilkan DataFrame (Gunakan Kurung Siku Ganda [[ ... ]])
kolom_pilihan = df[['Nama_Pelanggan', 'Kota', 'Total_Belanja_Juta']]
kolom_pilihan.head(3)
```

### B. Perbedaan Penting: `.loc` vs `.iloc`
- **`.loc[baris, kolom]`**: Memilih berdasarkan **Nama Label** (nama kolom atau indeks teks).
- **`.iloc[baris, kolom]`**: Memilih berdasarkan **Nomor Urut Angka** (*integer index*, mulai dari 0).

```python
# .iloc: Ambil baris ke-0 sampai ke-2, dan kolom ke-2 sampai ke-4 berdasarkan nomor urut angka
print("Contoh .iloc (Nomor Urut Angka):")
print(df.iloc[0:3, 2:5])

# .loc: Ambil baris berlabel 0 sampai 2, dan sebutkan nama kolomnya secara eksplisit
print("\nContoh .loc (Nama Kolom Eksplisit):")
print(df.loc[0:2, ['Nama_Pelanggan', 'Kategori_Produk', 'Total_Belanja_Juta']])
```

### C. Menyaring Baris dengan Syarat Logika (*Boolean Filtering*)

```python
# Filter 1 Syarat: Belanja > 10.0 Juta (Pembelian Bernilai Besar)
belanja_besar = df[df['Total_Belanja_Juta'] > 10.0]
print("Transaksi Belanja > Rp 10 Juta:")
print(belanja_besar[['Nama_Pelanggan', 'Kota', 'Kategori_Produk', 'Total_Belanja_Juta']])

# Filter Multi Syarat: Kota Jakarta DAN Metode Bayar Kartu Kredit
pelanggan_vip = df[(df['Kota'] == 'Jakarta') & (df['Metode_Bayar'] == 'Kartu Kredit')]
print("\nPelanggan Jakarta yang menggunakan Kartu Kredit:")
print(pelanggan_vip[['Nama_Pelanggan', 'Kota', 'Total_Belanja_Juta', 'Metode_Bayar']])
```

## 🛠️ 6. Transformasi Tabel: Tambah, Hapus, Urutkan & GroupBy

### A. Menambah Kolom Baru & Menghapus Kolom (`.drop()`)

```python
# 1. Menambah Kolom Baru: PPN 11% dan Total Bayar Akhir
df['Pajak_PPN_Juta'] = df['Total_Belanja_Juta'] * 0.11
df['Total_Bayar_Akhir'] = df['Total_Belanja_Juta'] + df['Pajak_PPN_Juta']

# 2. Menghapus Kolom yang Tidak Relevan: ID_Transaksi
df_bersih = df.drop(columns=['ID_Transaksi'])
print("Tabel setelah kolom ID_Transaksi dihapus:")
df_bersih.head(3)
```

### B. Mengurutkan Data (`.sort_values()`)

```python
# Mengurutkan transaksi berdasarkan Total Belanja dari yang TERBESAR ke TERKECIL (Descending)
df_terurut = df.sort_values(by='Total_Belanja_Juta', ascending=False)
print("Top 3 Transaksi Penjualan Terbesar:")
df_terurut[['Nama_Pelanggan', 'Kota', 'Kategori_Produk', 'Total_Belanja_Juta']].head(3)
```

### C. Mengelompokkan Data dengan `.groupby()` (Sangat Populer!)

```python
# 1. Rekapitulasi Total Omset dan Rata-rata per Kategori Produk
rekap_produk = df.groupby('Kategori_Produk').agg({
    'Total_Belanja_Juta': ['sum', 'mean'],
    'ID_Transaksi': 'count'
}).reset_index()

rekap_produk.columns = ['Kategori_Produk', 'Total_Omset_Juta', 'Rata_Rata_Juta', 'Unit_Terjual']
print("=== REKAP PENJUALAN PER KATEGORI PRODUK ===")
print(rekap_produk.sort_values(by='Total_Omset_Juta', ascending=False))

# 2. Rekapitulasi Omset Penjualan per Bulan
rekap_bulan = df.groupby('Bulan')['Total_Belanja_Juta'].sum().reset_index()
print("\n=== TREN OMSET PER BULAN ===")
print(rekap_bulan)
```

### D. Menyimpan Hasil Rekapitulasi ke File CSV Baru (`.to_csv()`)

```python
# Menyimpan tabel hasil analisis ke file CSV baru
rekap_produk.to_csv('rekap_produk_terlaris.csv', index=False)
print("File 'rekap_produk_terlaris.csv' berhasil disimpan ke disk! 💾")
```

## 📊 7. Visualisasi Data Lengkap (5 Jenis Diagram Utama)

Mari kita sajikan analisis kita dalam 5 diagram visual yang memukau:
1. **Bar Plot**: Total omset per kategori produk.
2. **Line Chart**: Tren pertumbuhan omset per bulan.
3. **Histogram**: Sebaran sebaran usia pelanggan.
4. **Scatter Plot**: Hubungan antara usia vs total belanja.
5. **Heatmap Korelasi**: Memetakan hubungan seluruh kolom numerik.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 1. Diagram Batang (Bar Plot): Total Omset per Kategori Produk
plt.figure(figsize=(7, 4))
sns.barplot(data=rekap_produk.sort_values(by='Total_Omset_Juta', ascending=False), x='Kategori_Produk', y='Total_Omset_Juta', palette='Blues_r')
plt.title('Total Omset Berdasarkan Kategori Produk (Juta Rp)', fontweight='bold')
plt.ylabel('Omset (Juta Rp)')
plt.show()
```

```python
# 2. Diagram Garis (Line Chart): Tren Pertumbuhan Omset per Bulan
plt.figure(figsize=(7, 3.5))
sns.lineplot(data=rekap_bulan, x='Bulan', y='Total_Belanja_Juta', marker='o', color='crimson', linewidth=2.5, markersize=8)
plt.title('Tren Pertumbuhan Omset Toko per Bulan (Jan - Apr)', fontweight='bold')
plt.ylabel('Total Omset (Juta Rp)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
# 3. Histogram: Sebaran Usia Pelanggan
plt.figure(figsize=(6, 3.5))
sns.histplot(data=df, x='Usia', bins=5, kde=True, color='teal')
plt.title('Distribusi Sebaran Usia Pelanggan (Histogram & KDE)', fontweight='bold')
plt.xlabel('Usia (Tahun)')
plt.ylabel('Jumlah Orang')
plt.show()
```

```python
# 4. Diagram Titik (Scatter Plot): Hubungan Usia vs Total Belanja
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x='Usia', y='Total_Belanja_Juta', hue='Kategori_Produk', style='Metode_Bayar', s=130)
plt.title('Hubungan Usia Pelanggan vs Total Belanja', fontweight='bold')
plt.xlabel('Usia (Tahun)')
plt.ylabel('Total Belanja (Juta Rp)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
# 5. Heatmap Korelasi: Hubungan Seluruh Kolom Angka
kolom_angka = df[['Usia', 'Total_Belanja_Juta', 'Pajak_PPN_Juta', 'Total_Bayar_Akhir']]

plt.figure(figsize=(6, 4))
sns.heatmap(kolom_angka.corr(), annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Matriks Korelasi Antar Variabel Numerik', fontweight='bold')
plt.show()
```
