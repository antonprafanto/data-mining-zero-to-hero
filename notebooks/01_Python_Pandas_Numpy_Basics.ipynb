## 🧱 2. Python Esensial: Variabel & Struktur Data

### A. Variabel & Tipe Primitif
Bayangkan **Variabel** seperti kardus penyimpanan yang Anda beri label nama:
- **String (Teks)** : Harus diapit tanda petik (`"Jakarta"`)
- **Integer (Bilangan Bulat)** : Angka tanpa koma (`28`)
- **Float (Bilangan Desimal)** : Angka pecahan dengan titik (`7.5`)
- **Boolean (Logika)** : Hanya bernilai Benar (`True`) atau Salah (`False`)

```python
# Menyimpan data pelanggan ke variabel sederhana
nama_pelanggan = "Budi Santoso"      # String (Teks)
usia = 29                           # Integer (Angka Bulat)
total_belanja_juta = 4.75           # Float (Angka Desimal)
member_aktif = True                 # Boolean (True/False)

print(f"Pelanggan : {nama_pelanggan}")
print(f"Usia      : {usia} tahun (Tipe: {type(usia).__name__})")
print(f"Belanja   : Rp {total_belanja_juta} Juta (Tipe: {type(total_belanja_juta).__name__})")
print(f"Member    : {member_aktif} (Tipe: {type(member_aktif).__name__})")
```

### B. Dua Wadah Data Paling Sering Dipakai: List & Dictionary

1. **List `[ ... ]`**: Wadah daftar berurut seperti daftar belanjaan. Urutannya dihitung mulai dari indeks **0**!
2. **Dictionary `{ kunci : nilai }`**: Wadah berpasangan seperti buku kontak telepon.

```python
# 1. Contoh List (Daftar Kota Cabang Toko)
daftar_kota = ['Jakarta', 'Surabaya', 'Bandung', 'Medan']
print(f"Daftar Kota        : {daftar_kota}")
print(f"Kota Pertama (idx 0): {daftar_kota[0]}")  # Indeks dimulai dari 0!
print(f"Kota Ketiga  (idx 2): {daftar_kota[2]}")

# 2. Contoh Dictionary (Data Profil Nasabah)
profil_nasabah = {
    'nama': 'Citra Lestari',
    'pekerjaan': 'Wiraswasta',
    'penghasilan_juta': 12.5,
    'skor_kredit': 750
}
print(f"\nNama Nasabah       : {profil_nasabah['nama']}")
print(f"Penghasilan        : Rp {profil_nasabah['penghasilan_juta']} Juta")
```

## ⚡ 3. Mengenal NumPy: Mesin Kalkulator Angka Super Cepat

### Kenapa Kita Butuh NumPy?
Jika Anda memiliki daftar harga 1.000 barang di Excel dan ingin memberikan diskon 10%, di Python biasa Anda harus membuat perulangan (*loop*) yang lambat.  
Dengan **NumPy Array**, Anda bisa mengalikan seluruh 1.000 angka tersebut secara sekaligus dalam satu kedipan mata (*Vectorized Operation*)!

```python
import numpy as np

# Membuat array harga belanjaan dalam ribuan rupiah
harga = np.array([50, 100, 150, 200, 250])  # [50rb, 100rb, ...]

# Memberikan diskon 10% (artinya harga menjadi 90% dari harga asli)
harga_diskon = harga * 0.9

print(f"Harga Asli   : {harga}")
print(f"Harga Diskon : {harga_diskon}")

# Fungsi Statistik NumPy yang Sering Dipakai:
print("\n--- Statistik Cepat NumPy ---")
print(f"Rata-rata (Mean)      : Rp {np.mean(harga)} Ribu")
print(f"Nilai Tengah (Median) : Rp {np.median(harga)} Ribu")
print(f"Nilai Terendah (Min)  : Rp {np.min(harga)} Ribu")
print(f"Nilai Tertinggi (Max) : Rp {np.max(harga)} Ribu")
print(f"Standar Deviasi (Std) : {np.std(harga):.2f}")
```

## 🐼 4. Jantung Data Mining: Membedah Pandas DataFrame

Bayangkan **Pandas** seperti program Microsoft Excel yang super bertenaga. Di Pandas ada dua konsep inti:
1. **Series**: Satu kolom data tunggal (1 Dimensi).
2. **DataFrame**: Kumpulan kolom yang membentuk lembar tabel utuh (2 Dimensi: ada baris dan kolom).

Mari kita buat tabel transaksi toko ritel pertama kita:

```python
import pandas as pd

# Membuat DataFrame dari Dictionary gabungan List
data_toko = {
    'ID_Transaksi': [101, 102, 103, 104, 105, 106, 107, 108],
    'Nama_Pelanggan': ['Andi', 'Budi', 'Citra', 'Dewi', 'Eko', 'Fani', 'Gita', 'Hadi'],
    'Kota': ['Jakarta', 'Surabaya', 'Bandung', 'Jakarta', 'Surabaya', 'Medan', 'Bandung', 'Jakarta'],
    'Usia': [25, 34, 29, 42, 19, 31, 23, 38],
    'Total_Belanja_Juta': [1.5, 4.2, 0.8, 7.5, 1.2, 3.4, 2.1, 5.0],
    'Metode_Bayar': ['QRIS', 'Transfer', 'Tunai', 'Kartu Kredit', 'QRIS', 'Transfer', 'QRIS', 'Kartu Kredit']
}

df = pd.DataFrame(data_toko)
print("=== TABEL DATA TRANSAKSI TOKO RITEL ===")
df
```

### 🔍 5 Perintah Sakti untuk Inspeksi Tabel Pertama Kali:
Setiap kali Anda membuka dataset baru, selalu jalankan perintah-perintah ini untuk mengenali karakter data:

```python
# 1. Dimensi data: (Jumlah Baris, Jumlah Kolom)
print(f"Ukuran Tabel (Baris, Kolom): {df.shape}")

# 2. Daftar nama-nama kolom
print(f"Nama Kolom: {df.columns.tolist()}\n")

# 3. .info() : Mengetahui tipe data dan apakah ada data kosong
print("--- Informasi Tipe Data (.info()) ---")
df.info()
```

```python
# 4. .describe() : Ringkasan statistik cepat untuk seluruh kolom angka
print("--- Ringkasan Statistik Otomatis (.describe()) ---")
df.describe()
```

## 🎯 5. Seni Memilih & Menyaring Data (Slicing & Filtering)

Di Excel Anda biasa menggunakan tombol Filter. Di Pandas, Anda bisa menyaring jutaan baris data hanya dengan satu baris kode rapi!

### A. Memilih Kolom Tertentu

```python
# Memilih 1 kolom saja (Hasilnya adalah Series)
kolom_nama = df['Nama_Pelanggan']

# Memilih beberapa kolom sekaligus (Gunakan tanda kurung siku ganda [[ ... ]])
kolom_terpilih = df[['Nama_Pelanggan', 'Kota', 'Total_Belanja_Juta']]
kolom_terpilih.head(4)
```

### B. Menyaring Baris Berdasarkan Syarat (*Filtering*)
Misalnya, kita ingin mencari transaksi yang belanjanya di atas 3 Juta Rupiah:

```python
# Filter 1 Syarat: Belanja > 3 Juta
belanja_besar = df[df['Total_Belanja_Juta'] > 3.0]
print("Pelanggan dengan Belanja > Rp 3 Juta:")
belanja_besar[['Nama_Pelanggan', 'Kota', 'Total_Belanja_Juta']]
```

```python
# Filter 2 Syarat: Berasal dari 'Jakarta' DAN Metode Bayar 'Kartu Kredit'
# Gunakan tanda & (DAN) atau | (ATAU)
pelanggan_vip = df[(df['Kota'] == 'Jakarta') & (df['Metode_Bayar'] == 'Kartu Kredit')]
print("Pelanggan Jakarta yang menggunakan Kartu Kredit:")
pelanggan_vip
```

---

## 🛠️ 6. Transformasi Tabel: Tambah Kolom, Hapus Kolom & GroupBy

### A. Menambah Kolom Baru
Misalnya pemerintah menetapkan PPN 11% untuk setiap transaksi:

```python
# Menghitung PPN 11%
df['Pajak_PPN_Juta'] = df['Total_Belanja_Juta'] * 0.11

# Menghitung Total Bayar Akhir (Belanja + PPN)
df['Total_Bayar_Akhir'] = df['Total_Belanja_Juta'] + df['Pajak_PPN_Juta']

# Mengkategorikan umur: Muda (<30) vs Dewasa (>=30)
df['Kategori_Usia'] = df['Usia'].apply(lambda x: 'Muda' if x < 30 else 'Dewasa')

df[['Nama_Pelanggan', 'Usia', 'Kategori_Usia', 'Total_Belanja_Juta', 'Total_Bayar_Akhir']]
```

### B. Mengelompokkan Data dengan `.groupby()` (Sangat Populer!)
Berapa rata-rata belanja dan total omset di setiap kota?

```python
# Mengelompokkan data berdasarkan Kota
rekap_kota = df.groupby('Kota').agg({
    'Total_Belanja_Juta': ['mean', 'sum'],
    'ID_Transaksi': 'count'
}).reset_index()

# Merapikan nama kolom hasil agregasi
rekap_kota.columns = ['Kota', 'Rata_Rata_Belanja_Juta', 'Total_Omset_Juta', 'Jumlah_Transaksi']
print("=== REKAP PENJUALAN PER KOTA ===")
rekap_kota
```

## 📊 7. Visualisasi Data: Menemukan Pola Lewat Gambar (Seaborn & Matplotlib)

Tabel angka sering kali membosankan bagi atasan atau klien. Mari kita ubah tabel di atas menjadi grafik visual yang memukau!

### Panduan Singkat: Kapan Memakai Grafik Apa?
- **Bar Plot (Diagram Batang)** : Membandingkan angka antar kategori (contoh: Total omset per kota).
- **Histogram** : Melihat sebaran frekuensi data numerik (contoh: Sebaran usia pelanggan).
- **Scatter Plot (Diagram Titik)** : Melihat keterkaitan antara 2 variabel angka (contoh: Usia vs Total Belanja).

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Atur gaya tampilan visual
sns.set_theme(style="whitegrid")

# 1. Diagram Batang: Total Omset per Kota
plt.figure(figsize=(8, 4))
sns.barplot(
    data=rekap_kota.sort_values(by='Total_Omset_Juta', ascending=False),
    x='Kota',
    y='Total_Omset_Juta',
    palette='Blues_r'
)
plt.title('Total Omset Penjualan Berdasarkan Kota (Juta Rp)', fontsize=13, fontweight='bold')
plt.xlabel('Kota Cabang', fontsize=11)
plt.ylabel('Omset (Juta Rp)', fontsize=11)
plt.show()
```

```python
# 2. Diagram Titik (Scatter Plot): Hubungan Usia vs Total Belanja
plt.figure(figsize=(7, 4.5))
sns.scatterplot(
    data=df,
    x='Usia',
    y='Total_Belanja_Juta',
    hue='Metode_Bayar',
    style='Kategori_Usia',
    s=120
)
plt.title('Hubungan Usia Pelanggan vs Total Belanja', fontsize=13, fontweight='bold')
plt.xlabel('Usia (Tahun)', fontsize=11)
plt.ylabel('Total Belanja (Juta Rp)', fontsize=11)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

💡 **Wawasan Bisnis dari Grafik di Atas:**  
Perhatikan bahwa pelanggan dengan usia di atas 35 tahun cenderung melakukan belanja bernilai tinggi (> Rp 5 Juta) dan memilih metode pembayaran **Kartu Kredit**!  
Wawasan sederhana ini langsung bisa digunakan tim pemasaran untuk membuat promo kartu kredit khusus usia dewasa!
