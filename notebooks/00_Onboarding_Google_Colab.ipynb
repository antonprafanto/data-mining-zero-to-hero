## ⛏️ 2. Apa Itu Data Mining Sebenarnya?

### 🌊 Analogi: Mendulang Emas di Sungai Keruh
Bayangkan Anda berdiri di tepi sungai yang airnya keruh berlumpur:
1. **Pasir & Lumpur Sungai** = **Data Mentah (*Raw Data*)**. Di dunia nyata, ini adalah jutaan struk kasir Indomaret, rekam medis pasien di rumah sakit, atau riwayat transaksi perbankan yang menumpuk tak beraturan.
2. **Dulang & Ayakan** = **Algoritma Data Mining**. Alat dan teknik untuk menyaring, memisahkan kotoran, dan menyortir butiran pasir.
3. **Butiran Emas Murni** = **Pengetahuan Berharga (*Knowledge / Insight*)**. Informasi tersembunyi yang bernilai tinggi. Contoh: *"Ternyata nasabah yang memiliki saldo tabungan stabil tetapi meminjam uang di atas jam 11 malam memiliki risiko gagal bayar 4x lebih tinggi!"*.

### 🔍 Perbedaan Cepat: AI vs Machine Learning vs Data Mining vs Data Science
| Bidang | Fokus Utama | Pertanyaan yang Dijawab | Contoh Kasus |
| :--- | :--- | :--- | :--- |
| **Data Mining** | Menemukan pola tersembunyi dari sekumpulan data besar (*Pattern Discovery*). | *"Pola atau kebiasaan apa yang tersembunyi di data masa lalu kita?"* | Menemukan barang apa yang sering dibeli bersamaan oleh pelanggan swalayan. |
| **Machine Learning** | Melatih algoritma agar bisa memprediksi masa depan tanpa diprogram manual. | *"Berapa kemungkinan nasabah ini akan macet bayar bulan depan?"* | Sistem credit scoring otomatis di perbankan. |
| **Artificial Intelligence (AI)** | Meniru kecerdasan manusia secara luas (teks, suara, visual). | *"Bisakah komputer merespons percakapan selayaknya manusia?"* | ChatGPT, asisten suara, mobil otonom. |
| **Data Science** | Payung besar yang menggabungkan statistika, pemrograman, dan strategi bisnis. | *"Bagaimana strategi perusahaan memanfaatkan seluruh aset data?"* | Perencanaan strategi digital korporasi berbasis data. |

## 🔄 3. Alur Kerja Standar Dunia: CRISP-DM

Praktisi data mining di perusahaan global (Google, Netflix, Gojek, Bank Mandiri) tidak bekerja asal ketik kode. Mereka mengikuti siklus **CRISP-DM** (*Cross-Industry Standard Process for Data Mining*):

1. **Business Understanding (Pahami Masalah)**: Tanyakan dulu masalah bisnisnya. *Contoh: Mengapa omset toko kita turun 20% bulan ini?*
2. **Data Understanding (Pahami Datanya)**: Kumpulkan datanya, cek ada berapa baris, kolom apa saja yang tersedia, apakah banyak yang kosong?
3. **Data Preparation (Bersihkan Datanya)**: 80% waktu Anda akan berada di sini! Buang data salah ketik, isi data yang bolong, dan samakan formatnya.
4. **Modeling (Latih Algoritma)**: Masukkan data bersih ke algoritma (seperti Decision Tree, K-Means, atau Random Forest).
5. **Evaluation (Uji Kualitas)**: Periksa apakah tebakan model akurat dan benar-benar menjawab masalah pada langkah ke-1.
6. **Deployment (Sajikan ke Pengguna)**: Buatkan antarmuka web interaktif (misalnya dengan Streamlit) agar hasil model bisa dipakai langsung oleh bos atau pelanggan!

## 🌸 9. Cicipan Pertama: Mengintip Dataset Bunga Iris (The "Hello World" of Data Mining)

Untuk membuktikan betapa menyenangkannya data mining, mari kita baca dataset legendaris bunga Iris dalam 2 baris kode saja:

```python
# Memuat dataset bunga Iris bawaan pustaka Seaborn
df_iris = sns.load_dataset('iris')

# Tampilkan 5 baris pertama data
df_iris.head()
```

Mari kita bedah tabel di atas dengan bahasa awam:
- **`sepal_length` & `sepal_width`**: Panjang dan lebar kelopak pelindung bunga (dalam cm).
- **`petal_length` & `petal_width`**: Panjang dan lebar mahkota bunga (dalam cm).
  *(Keempat kolom ini disebut **Fitur / Features** = Ciri-ciri fisik bunga)*.
- **`species`**: Jenis spesies bunga (Setosa, Versicolor, Virginica).
  *(Kolom ini disebut **Target / Label** = Kunci jawaban yang ingin kita tebak)*.

Sekarang, mari kita minta komputer menghitung ringkasan statistik (rata-rata, nilai terkecil, dan terbesar) secara instan:

```python
# Ringkasan statistik cepat dari angka-angka bunga
df_iris.describe()
```

Dan mari kita buat grafik sebarannya dengan 3 baris kode visual:

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 4))
sns.scatterplot(data=df_iris, x='petal_length', y='petal_width', hue='species', palette='Dark2', s=80)
plt.title('Sebaran Ukuran Mahkota Bunga Iris Berdasarkan Spesies')
plt.xlabel('Panjang Mahkota / Petal (cm)')
plt.ylabel('Lebar Mahkota / Petal (cm)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

💡 **Lihat grafiknya!**  
Bahkan sebelum kita mempelajari rumus machine learning, mata kita sudah bisa melihat polanya secara alami:
- Bunga **Setosa** (titik hijau tua) berkelompok di kiri bawah dengan mahkota yang sangat kecil.
- Bunga **Versicolor** berada di tengah.
- Bunga **Virginica** berada di kanan atas dengan mahkota paling besar.

**Itulah esensi Data Mining!** Menemukan pola pembeda yang jelas dari tumpukan angka-angka mentah!
