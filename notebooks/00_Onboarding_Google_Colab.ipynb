## ⛏️ 2. Apa Sebenarnya Data Mining Itu?

### 🌊 Analogi Dunia Nyata: Mendulang Emas di Sungai Keruh
Bayangkan Anda sedang berada di sungai yang airnya keruh berlumpur:
1. **Pasir & Lumpur Sungai** = **Data Mentah (*Raw Data*)**. Ini adalah tumpukan jutaan data berantakan: struk kasir minimarket, rekam medis pasien di puskesmas, atau log histori belanja e-commerce yang tersimpan di server.
2. **Dulang & Saringan Kawat** = **Algoritma Data Mining**. Alat dan teknik untuk menyaring kotoran, membuang batu kerikil tak berguna, dan menyortir butiran pasir.
3. **Butiran Emas Murni** = **Wawasan Berharga (*Actionable Knowledge*)**. Penemuan pola tersembunyi yang bernilai tinggi.  
   *Contoh Emas Nyata:*  
   *Sebuah rumah sakit menemukan pola bahwa pasien diabetes yang rutin minum obat di jam 7 pagi memiliki peluang komplikasi 60% lebih rendah dibanding yang minum obat tidak beraturan.* Informasi seperti ini menyelamatkan nyawa dan menghasilkan keputusan berharga!

### 🔍 Kamus Cepat: Perbedaan Data Mining, Machine Learning, AI, dan Data Science
Agar Anda tidak bingung saat membaca istilah-istilah di internet:
- **Data Mining**: Fokus pada **menemukan pola tersembunyi** dari kumpulan data besar di masa lalu (*Pattern Discovery*).
- **Machine Learning**: Fokus pada **melatih mesin agar bisa memprediksi masa depan** tanpa harus diprogram secara manual langkah demi langkah (*Prediction*).
- **Artificial Intelligence (AI)**: Konsep payung besar tentang mesin cerdas yang mampu meniru perilaku manusia (robotika, ChatGPT, pengenal wajah).
- **Data Science**: Disiplin ilmu menyeluruh yang menggabungkan statistika, pemrograman, dan strategi bisnis untuk memecahkan masalah korporasi.

## 🔄 3. Alur Kerja Standar Industri Dunia: CRISP-DM

Setiap proyek data mining profesional di dunia selalu mengikuti 6 tahapan siklus **CRISP-DM** (*Cross-Industry Standard Process for Data Mining*):

```
   [1. Pahami Masalah]  ──►  [2. Pahami Data]   ──►  [3. Bersihkan Data]
   Business Understanding      Data Understanding      Data Preparation
                                                              │
                                                              ▼
   [6. Pasang ke Web]   ◄──  [5. Uji Kualitas]  ◄──  [4. Latih Algoritma]
       Deployment                Evaluation                 Modeling
```

1. **Business Understanding (Pahami Masalah Bisnis)**: Apa yang mau diselesaikan? *(Contoh: Mengapa banyak nasabah pinjol telat bayar cicilan?)*
2. **Data Understanding (Pahami Datanya)**: Kumpulkan datanya, cek ada berapa kolom, apa isinya, apakah ada yang salah ketik?
3. **Data Preparation (Bersihkan Datanya)**: **80% waktu praktisi data dihabiskan di sini!** Mengisi data kosong, membuang angka tidak masuk akal, dan menyamakan format.
4. **Modeling (Latih Algoritma)**: Memasukkan data bersih ke algoritma pembuat pola (seperti Decision Tree atau K-Means).
5. **Evaluation (Uji Kualitas Model)**: Menguji apakah tebakan model benar-benar tepat atau hanya kebetulan.
6. **Deployment (Sajikan ke Pengguna Nyata)**: Mengubah model menjadi aplikasi web interaktif (Streamlit) yang bisa diakses orang lain lewat browser.

## 🌸 9. Cicipan Praktik Pertama: Menemukan Pola pada Bunga Iris

Untuk merasakan keajaiban data mining secara langsung, mari kita buka dataset klasik dunia data science: **Bunga Iris**.

Jalankan sel di bawah ini untuk melihat 5 baris pertama datanya:

````python
# Membaca dataset bunga Iris bawaan pustaka Seaborn
df_iris = sns.load_dataset('iris')

# Menampilkan 5 baris pertama data
df_iris.head()
````

### 📖 Membedah Tabel di Atas dengan Bahasa Manusia:
Tabel di atas berisi pengukuran fisik 150 tangkai bunga:
- **`sepal_length` & `sepal_width`**: Panjang dan lebar kelopak pelindung daun bunga (dalam cm).
- **`petal_length` & `petal_width`**: Panjang dan lebar mahkota bunga yang berwarna (dalam cm).  
  👉 Keempat kolom angka ini disebut **Fitur (*Features*)** = Ciri-ciri fisik yang kita amati.
- **`species`**: Jenis varietas bunga (ada 3 jenis: *setosa*, *versicolor*, dan *virginica*).  
  👉 Kolom ini disebut **Target (*Label*)** = Kunci jawaban yang ingin ditebak oleh komputer.

Sekarang, mari kita buat diagram sebaran titik (*scatter plot*) untuk melihat apakah mahkota bunga (*petal*) bisa membedakan ketiga jenis bunga tersebut:

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4.5))
sns.scatterplot(
    data=df_iris,
    x='petal_length',
    y='petal_width',
    hue='species',
    palette='Dark2',
    s=90
)

plt.title('Pola Ukuran Mahkota Bunga Iris Berdasarkan Spesies', fontsize=13, fontweight='bold')
plt.xlabel('Panjang Mahkota / Petal (cm)', fontsize=11)
plt.ylabel('Lebar Mahkota / Petal (cm)', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
````

### 💡 Perhatikan Keajaiban Pola di Grafik Tersebut!
Bahkan tanpa rumus matematika yang memusingkan, mata Anda bisa melihat polanya dengan sangat jelas:
1. **Spesies Setosa (Titik Hijau Tua)**: Mengelompok rapi di pojok kiri bawah. Mahkotanya selalu berukuran mungil (panjang < 2 cm dan lebar < 0.6 cm).
2. **Spesies Versicolor (Titik Oranye)**: Berada di kelompok tengah.
3. **Spesies Virginica (Titik Ungu)**: Mengelompok di kanan atas dengan mahkota yang paling besar.

**Inilah hakikat dari Data Mining!**  
Kita mengajarkan komputer untuk mengenali batas-batas pemisah tersebut secara otomatis, sehingga jika besok ada turis menemukan bunga baru di hutan, komputer langsung tahu jenis bunganya hanya dengan mengukur panjang dan lebar mahkotanya!
