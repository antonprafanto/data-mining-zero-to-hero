## ⛏️ 2. Apa Sebenarnya Data Mining Itu?

### 🌊 Analogi: Mendulang Emas di Sungai Keruh
Istilah *Data Mining* (Penambangan Data) lahir dari analogi penambang emas tradisional:
1. **Pasir & Lumpur Keruh Sungai** = **Data Mentah (*Raw Data*)**. Ini adalah tumpukan jutaan data berantakan: struk kasir minimarket, rekam medis pasien di puskesmas, atau log histori belanja e-commerce yang tersimpan di database.
2. **Dulang & Saringan Kawat** = **Algoritma Data Mining**. Alat dan teknik untuk menyaring kotoran, membuang batu kerikil tak berguna, dan menyortir butiran pasir.
3. **Butiran Emas Murni** = **Wawasan Berharga (*Actionable Knowledge*)**. Penemuan pola tersembunyi yang bernilai tinggi.

> 💡 *"Data rich, but information poor."* (Kaya akan data mentah, tetapi miskin informasi).  
> Perusahaan saat ini kebanjiran data gigabyte setiap detik, namun tidak tahu maknanya. Tugas Anda sebagai *data miner* adalah mengubah tumpukan angka mati itu menjadi keputusan cerdas!

### 🔍 Kamus Cepat: Perbedaan Data Mining, Machine Learning, AI, dan Data Science
| Bidang | Fokus Utama | Pertanyaan yang Dijawab | Contoh Kasus |
| :--- | :--- | :--- | :--- |
| **Data Mining** | Menemukan pola tersembunyi dari sekumpulan data besar di masa lalu (*Pattern Discovery*). | *"Pola atau kebiasaan apa yang tersembunyi di data kita?"* | Menemukan produk apa yang sering dibeli bersamaan oleh pelanggan. |
| **Machine Learning** | Melatih algoritma agar bisa memprediksi masa depan tanpa diprogram manual langkah demi langkah. | *"Berapa peluang nasabah ini akan macet bayar bulan depan?"* | Sistem *credit scoring* otomatis di perbankan. |
| **Artificial Intelligence (AI)** | Meniru kecerdasan manusia secara menyeluruh (teks, suara, visual). | *"Bisakah komputer merespons percakapan selayaknya manusia?"* | ChatGPT, asisten suara Siri, mobil otonom. |
| **Data Science** | Payung besar yang menggabungkan statistika, pemrograman, dan strategi bisnis. | *"Bagaimana strategi perusahaan memanfaatkan seluruh aset data?"* | Perencanaan strategi digital korporasi berbasis data. |

## 🎯 3. Dua Tugas Besar Data Mining: Prediktif vs Deskriptif

Seluruh teknik dalam data mining di dunia selalu terbagi ke dalam **2 payung tugas utama**:

### A. Tugas Prediktif (*Predictive Tasks*)
- **Tujuan**: Memprediksi nilai masa depan yang belum diketahui berdasarkan data masa lalu.
- **Ada Target/Label yang ingin ditebak (*Supervised*)**.
- **Cabang Utama**:
  1. **Klasifikasi (*Classification*)**: Menebak label kategori diskrit (Contoh: Menentukan email SPAM atau BUKAN SPAM; Menentukan pasien POSITIF COVID atau NEGATIF).
  2. **Regresi (*Regression*)**: Menebak angka kontinu (Contoh: Menaksir harga jual rumah tahun depan; Memprediksi omset penjualan bulan depan).

### B. Tugas Deskriptif (*Descriptive Tasks*)
- **Tujuan**: Menemukan pola alami dan meringkas karakteristik data yang sudah ada agar mudah dipahami manusia.
- **Tidak ada kunci jawaban/target (*Unsupervised*)**.
- **Cabang Utama**:
  1. **Klasterisasi (*Clustering*)**: Mengelompokkan data berdasarkan kemiripan karakteristik (Contoh: Membagi pelanggan mall menjadi 3 kelompok: Pelanggan Hemat, Pelanggan Royal, Pelanggan Promo).
  2. **Aturan Asosiasi (*Association Rules*)**: Menemukan hubungan keterkaitan antar item (Contoh: *Market Basket Analysis* - pembeli popok sering membeli kopi kaleng).
  3. **Deteksi Anomali (*Anomaly Detection*)**: Menemukan data aneh yang menyimpang dari pola umum (Contoh: Deteksi transaksi kartu kredit yang dicuri).

## 🛑 4. Kapan Boleh Pakai Data Mining & Kapan TIDAK PERLU?

Sebagai orang awam, hindari kesalahan umum pemula: *"Semua masalah ingin diselesaikan pakai Machine Learning / Data Mining"*.

Gunakan pedoman aturan emas ini:
- ❌ **TIDAK PERLU Data Mining**: Jika masalah bisa diselesaikan dengan rumus matematika pasti atau query database biasa.  
  *(Contoh: Menghitung total gaji bulanan karyawan = Gaji Pokok + Tunjangan. Ini rumus pasti di Excel/SQL, tidak butuh machine learning!).*
- ❌ **TIDAK PERLU Data Mining**: Jika jumlah datanya sangat sedikit (misal hanya 10 baris data). Pola tidak bisa digali dari data yang terlalu sedikit.
- ✅ **WAJIB Data Mining**: Ketika Anda memiliki **data besar**, polanya **tidak diketahui secara pasti (*unknown pattern*)**, dan hubungan antar variabel terlalu rumit untuk dirumuskan dengan logika manusia biasa (*rule-based*).  
  *(Contoh: Memprediksi dari 10.000 nasabah, siapa yang berisiko kabur tidak membayar cicilan berdasarkan 40 riwayat transaksi mereka).*

## 🔄 5. Dua Metodologi Standar: KDD vs CRISP-DM

Dalam literatur akademik dan industri, ada 2 standar proses yang wajib Anda kenal:

### A. Metodologi Akademik: KDD (*Knowledge Discovery in Databases*)
Ditemukan oleh Fayyad et al., KDD memandang data mining sebagai salah satu langkah inti dalam rangkaian proses panjang:
1. **Data Selection**: Memilih subset data relevan yang ingin diteliti.
2. **Data Cleaning & Preprocessing**: Membersihkan data dari derau (*noise*) dan data hilang.
3. **Data Transformation**: Mengubah format, skala, dan reduksi dimensi.
4. **Data Mining**: Menerapkan algoritma cerdas untuk mengekstrak pola.
5. **Pattern Evaluation & Interpretation**: Menginterpretasikan pola menjadi wawasan bagi manusia.

### B. Metodologi Industri Terapan: CRISP-DM
CRISP-DM adalah standar industri paling populer karena memasukkan unsur tujuan bisnis dan penyajian aplikasi:

```
   [1. Pahami Masalah]  ──►  [2. Pahami Data]   ──►  [3. Bersihkan Data]
   Business Understanding      Data Understanding      Data Preparation
                                                              │
                                                              ▼
   [6. Pasang ke Web]   ◄──  [5. Uji Kualitas]  ◄──  [4. Latih Algoritma]
       Deployment                Evaluation                 Modeling
```

## 📊 6. Fondasi Mutlak: Mengenal 4 Tipe Data Dunia Nyata

Bagi orang awam, data sering dianggap hanya "angka" dan "teks biasa". Namun di data mining, variabel data dibagi menjadi **4 tipe utama**. Memahami 4 tipe ini sejak hari pertama akan membuat seluruh modul berikutnya (pembersihan data dan machine learning) menjadi sangat mudah!

| Kategori | Tipe Data | Ciri-Ciri | Contoh Nyata | Perlakuan di Data Mining |
| :--- | :--- | :--- | :--- | :--- |
| **Angka (Numerik)** | **1. Kontinu** | Angka pecahan/desimal hasil pengukuran fisik, rentangnya kontinu. | Tinggi badan (165.4 cm), Suhu (36.8°C), Harga tiket ($71.28). | Menggunakan regresi, penskalaan (*StandardScaler*). |
| **Angka (Numerik)** | **2. Diskrit** | Angka bulat utuh hasil membilang/menghitung barang (tidak ada pecahan). | Jumlah anak (2 orang), Jumlah kamar tidur (3), Jumlah saudara (1). | Dihitung rata-rata atau fitur numerik diskrit. |
| **Kategori (Kualitatif)**| **3. Nominal** | Label/nama tanpa tingkatan atau peringkat (setara nilainya). | Jenis kelamin (L/P), Warna mobil (Merah, Hitam), Pelabuhan asal (C, Q, S). | Diubah menjadi angka biner dengan *One-Hot Encoding*. |
| **Kategori (Kualitatif)**| **4. Ordinal** | Label yang memiliki tingkatan atau hierarki yang jelas. | Kelas tiket (Kelas 1 > Kelas 2 > Kelas 3), Kepuasan (Buruk < Cukup < Puas). | Diberi peringkat angka urut dengan *Ordinal Encoding*. |

## 🌸 12. Hands-on 4: Cicipan Pola Visual Bunga Iris (*First Pattern Discovery*)

Untuk merasakan keajaiban penemuan pola data secara visual, mari kita buka dataset klasik bunga Iris dan gambar diagramnya:

````python
# Membaca dataset bunga Iris bawaan pustaka Seaborn
df_iris = sns.load_dataset('iris')

# Gambar diagram sebaran titik (Scatter Plot)
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4.5))
sns.scatterplot(
    data=df_iris,
    x='petal_length',
    y='petal_width',
    hue='species',
    palette='Dark2',
    s=90
)

plt.title('Pola Ukuran Mahkota Bunga Iris Berdasarkan Spesies', fontsize=13, fontweight='bold')
plt.xlabel('Panjang Mahkota / Petal (cm)', fontsize=11)
plt.ylabel('Lebar Mahkota / Petal (cm)', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
````

### 💡 Perhatikan Keajaiban Pola di Grafik Tersebut!
Bahkan tanpa rumus matematika yang memusingkan, mata Anda bisa melihat polanya dengan sangat jelas:
1. **Spesies Setosa (Titik Hijau Tua)**: Mengelompok rapi di pojok kiri bawah. Mahkotanya selalu berukuran mungil (panjang < 2 cm dan lebar < 0.6 cm).
2. **Spesies Versicolor (Titik Oranye)**: Berada di kelompok tengah.
3. **Spesies Virginica (Titik Ungu)**: Mengelompok di kanan atas dengan mahkota yang paling besar.
