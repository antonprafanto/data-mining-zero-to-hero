## ✍️ 3. Menulis File Aplikasi `app_demo.py` Menggunakan Magic Command `%%writefile`

```python
%%writefile app_demo.py
import streamlit as st
import pandas as pd
import numpy as np

st.title("🎉 Web App Data Mining Pertama Saya!")
st.write("Aplikasi ini dibuat langsung dari Google Colab menggunakan Streamlit.")

nama = st.text_input("Siapa nama Anda?", "Pembelajar Hero")
nilai = st.slider("Berapa tingkat semangat Anda belajar data mining? (1-100)", 1, 100, 95)

if st.button("Tampilkan Pesan Semangat!"):
    st.success(f"Halo **{nama}**! Semangat Anda berada di angka **{nilai}%**! Tetap konsisten dan jadilah Hero Data Mining! 🚀")

# Diagram sederhana
chart_data = pd.DataFrame(np.random.randn(20, 3), columns=['Akurasi Model', 'Presisi', 'Recall'])
st.line_chart(chart_data)
```
