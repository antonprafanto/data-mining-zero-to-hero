## 📖 Kamus Istilah Awam (Jargon Buster Modul 10)

---

### 🔄 Siklus Hidup Eksekusi Reaktif Streamlit (*Reactive Execution*)
Berbeda dengan web konvensional di mana halaman melakukan refresh penuh (*full page reload*), Streamlit bekerja secara **reaktif**:
```
 Pengguna Menggeser Slider Usia ➔ Streamlit Menjalankan Skrip dari Baris 1
                                                     │
                                                     ▼
 Variabel `usia` Otomatis Terisi Angka Baru ➔ Model Pipeline Menghitung Prediksi
                                                     │
                                                     ▼
 Grafik & Teks di Layar Terbarui Seketika (Nol Detik Reloading!)
```
Prinsip ini membuat Anda bisa menulis kode seperti menulis skrip analitik biasa dari atas ke bawah, tanpa perlu pusing mengatur *event listener* atau fungsi callback yang rumit!


## 🎛️ 2. Mengenal Komponen Antarmuka (*Input Widgets*) Streamlit

Streamlit menyediakan serangkaian komponen interaktif (*widgets*) yang secara otomatis memetakan aksi pengguna di browser menjadi variabel Python:

| Nama Komponen | Cuplikan Sintaks Kode | Tipe Nilai yang Dihasilkan | Contoh Skenario Penggunaan |
| :--- | :--- | :--- | :--- |
| **`st.slider()`** | `umur = st.slider("Usia:", 1, 80, 25)` | Integer / Float | Menentukan umur, tahun pengalaman, atau probabilitas batas. |
| **`st.selectbox()`** | `gender = st.selectbox("Gender:", ["Pria", "Wanita"])` | String terpilih | Memilih kategori diskrit (jenis kelamin, tipe kabin, kota). |

| **`st.file_uploader()`** | `file = st.file_uploader("Unggah CSV:", type=["csv"])` | File Buffer (Bytes) | Menerima berkas spreadsheet dari pengguna untuk batch processing. |
| **`st.button()`** | `if st.button("Hitung Prediksi"):` | Boolean (`True`/`False`) | Memicu eksekusi kalkulasi model hanya saat pengguna sudah siap. |

Semua widget ini dapat diletakkan di halaman utama dengan `st.` atau dipindahkan ke panel samping menggunakan `st.sidebar.` agar tampilan lebih rapi dan lapang!

## 🏭 4. Mempersiapkan Artefak Model Pipeline & Data Uji Batch

Sesuai filosofi kurikulum kita, modul ini adalah kelanjutan langsung dari **Modul 09**.  
Aplikasi web kita membutuhkan:
1. **File Model Biner (`model_prediksi_titanic_pipeline.joblib`)**: Memuat seluruh ban berjalan machine learning (*ColumnTransformer* + *RandomForestClassifier*) yang mampu menerima data mentah dengan nilai hilang (*missing values*), melakukan imputasi, standardisasi, encoding, dan memprediksi probabilitas keselamatan.
2. **File CSV Sampel (`penumpang_uji_batch.csv`)**: Tabel data 10 calon penumpang uji coba dengan profil beragam untuk menguji fitur prediksi massal (*Batch Prediction*).

Mari kita pastikan kedua berkas ini siap di media penyimpanan sebelum kita merakit antarmuka web:


````python
import os
import joblib
import numpy as np
import pandas as pd
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.ensemble import RandomForestClassifier

nama_file_model = "model_prediksi_titanic_pipeline.joblib"
nama_file_csv_batch = "penumpang_uji_batch.csv"

# 1. Pastikan Model Pipeline Titanic Tersedia
if not os.path.exists(nama_file_model):
    print("⚙️ Melatih dan merakit model pipeline Titanic untuk aplikasi web...")
    
    # Dataset pelatihan Titanic realistis
    np.random.seed(42)
    n_samples = 400
    
    data_train = pd.DataFrame({
        'Umur': np.random.choice([np.nan, 22.0, 38.0, 26.0, 35.0, 54.0, 2.0, 27.0, 14.0, 4.0, 58.0], size=n_samples),
        'Tarif_Tiket': np.random.exponential(scale=35.0, size=n_samples) + 7.25,
        'Jenis_Kelamin': np.random.choice(['pria', 'wanita'], size=n_samples, p=[0.65, 0.35]),
        'Kelas_Kabin': np.random.choice(['VIP', 'Bisnis', 'Ekonomi'], size=n_samples, p=[0.24, 0.21, 0.55])
    })
    
    # Logika keselamatan historis (prioritas wanita, anak-anak, dan kelas atas)
    skor_prob = (
        (data_train['Jenis_Kelamin'] == 'wanita') * 0.50 +
        (data_train['Kelas_Kabin'] == 'VIP') * 0.25 +
        (data_train['Kelas_Kabin'] == 'Bisnis') * 0.10 +
        (data_train['Umur'].fillna(28) < 12) * 0.20 +
        np.random.normal(0, 0.10, size=n_samples)
    )
    data_train['Selamat'] = (skor_prob >= 0.40).astype(int)
    
    fitur_numerik = ['Umur', 'Tarif_Tiket']
    fitur_kategorikal = ['Jenis_Kelamin', 'Kelas_Kabin']
    
    pipa_numerik = Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ])
    
    pipa_kategorikal = Pipeline([
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('encoder', OneHotEncoder(handle_unknown='ignore'))
    ])
    
    preprocessor = ColumnTransformer([
        ('num', pipa_numerik, fitur_numerik),
        ('cat', pipa_kategorikal, fitur_kategorikal)
    ])
    
    model_pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('classifier', RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42))
    ])
    
    X_train = data_train[fitur_numerik + fitur_kategorikal]
    y_train = data_train['Selamat']
    
    model_pipeline.fit(X_train, y_train)
    joblib.dump(model_pipeline, nama_file_model)
    print(f"✅ Model pipeline baru berhasil dilatih dan disimpan ke: {nama_file_model}")
else:
    print(f"✅ Model pipeline '{nama_file_model}' sudah tersedia dan siap digunakan!")

# 2. Buat File Sampel Penumpang untuk Uji Coba Batch CSV
df_sampel_batch = pd.DataFrame([
    {"Nama_Penumpang": "Rose DeWitt Bukater", "Umur": 17.0, "Tarif_Tiket": 150.00, "Jenis_Kelamin": "wanita", "Kelas_Kabin": "VIP"},
    {"Nama_Penumpang": "Jack Dawson", "Umur": 20.0, "Tarif_Tiket": 7.50, "Jenis_Kelamin": "pria", "Kelas_Kabin": "Ekonomi"},
    {"Nama_Penumpang": "Margaret Brown (Molly)", "Umur": 44.0, "Tarif_Tiket": 85.00, "Jenis_Kelamin": "wanita", "Kelas_Kabin": "VIP"},
    {"Nama_Penumpang": "Thomas Andrews", "Umur": 39.0, "Tarif_Tiket": 0.00, "Jenis_Kelamin": "pria", "Kelas_Kabin": "VIP"},
    {"Nama_Penumpang": "Anak Kecil Tanpa Tiket", "Umur": 5.0, "Tarif_Tiket": 12.00, "Jenis_Kelamin": "pria", "Kelas_Kabin": "Bisnis"},
    {"Nama_Penumpang": "Penumpang Lansia", "Umur": 71.0, "Tarif_Tiket": 34.00, "Jenis_Kelamin": "pria", "Kelas_Kabin": "Bisnis"},
    {"Nama_Penumpang": "Wanita Pekerja", "Umur": 24.0, "Tarif_Tiket": 15.50, "Jenis_Kelamin": "wanita", "Kelas_Kabin": "Ekonomi"},
    {"Nama_Penumpang": "Pria Pekerja Tambang", "Umur": 31.0, "Tarif_Tiket": 8.05, "Jenis_Kelamin": "pria", "Kelas_Kabin": "Ekonomi"},
    {"Nama_Penumpang": "Ibu Muda & Bayi", "Umur": 22.0, "Tarif_Tiket": 41.57, "Jenis_Kelamin": "wanita", "Kelas_Kabin": "Bisnis"},
    {"Nama_Penumpang": "Pengusaha Muda", "Umur": 30.0, "Tarif_Tiket": 110.88, "Jenis_Kelamin": "pria", "Kelas_Kabin": "VIP"}
])

df_sampel_batch.to_csv(nama_file_csv_batch, index=False)
print(f"✅ Berkas CSV sampel berhasil dibuat: {nama_file_csv_batch} ({len(df_sampel_batch)} baris profil)")
````

````python
# Simulasi Pengujian Model dan Visualisasi Plotly di Dalam Notebook
import joblib
import pandas as pd

# 1. Muat Model
model = joblib.load("model_prediksi_titanic_pipeline.joblib")

# 2. Sampel Data Penumpang Uji Coba (Wanita Kelas VIP Umur 25)
penumpang_tes = pd.DataFrame([{
    'Umur': 25.0,
    'Tarif_Tiket': 85.0,
    'Jenis_Kelamin': 'wanita',
    'Kelas_Kabin': 'VIP'
}])

# 3. Hitung Prediksi & Probabilitas
hasil_pred = model.predict(penumpang_tes)[0]
prob_selamat = model.predict_proba(penumpang_tes)[0][1]
prob_persen = prob_selamat * 100

print(f"=== HASIL SIMULASI IN-NOTEBOOK ===")
print(f"Profil Penumpang : Wanita 25 Tahun, Tiket VIP ($85.00)")
print(f"Status Prediksi  : {'SELAMAT 🎉' if hasil_pred == 1 else 'TIDAK SELAMAT ⚠️'}")
print(f"Peluang Selamat  : {prob_persen:.1f}%")

# 4. Bangun Speedometer Gauge Plotly (jika Plotly terpasang)
try:
    import plotly.graph_objects as go
    warna_bar = "#2ca02c" if prob_selamat >= 0.5 else "#d62728"
    fig_gauge = go.Figure(go.Indicator(
        mode="gauge+number",
        value=prob_persen,
        number={'suffix': "%", 'font': {'size': 32}},
        domain={'x': [0, 1], 'y': [0, 1]},
        title={'text': "Simulasi Indikator Peluang Selamat (Plotly Gauge)", 'font': {'size': 18}},
        gauge={
            'axis': {'range': [0, 100], 'tickwidth': 1, 'tickcolor': "gray"},
            'bar': {'color': warna_bar},
            'steps': [
                {'range': [0, 50], 'color': "rgba(214, 39, 40, 0.15)"},
                {'range': [50, 100], 'color': "rgba(44, 160, 44, 0.15)"}
            ],
            'threshold': {
                'line': {'color': "black", 'width': 3},
                'thickness': 0.75,
                'value': 50
            }
        }
    ))
    fig_gauge.update_layout(height=260, margin=dict(l=20, r=20, t=30, b=20))
    try:
        fig_gauge.show()
    except Exception:
        print("✅ Grafik Plotly Gauge berhasil dikonstruksi!")
except ImportError:
    print(f"📊 Indikator Peluang Selamat: {prob_persen:.1f}% (Plotly belum terinstal di lingkungan ini)")
````

4. **Tab 2: 📂 Prediksi Massal (*Batch Prediction*)**:

   * Memproses seluruh baris data menggunakan model pipeline Scikit-Learn.

   * Tombol unduh hasil prediksi lengkap ke format CSV.
5. **Tab 3: ℹ️ Arsitektur AI & CRISP-DM**: Edukasi bagi pengguna mengenai alur kerja data mining di balik layar.


````python
%%writefile app_colab.py
import streamlit as st
import pandas as pd
import numpy as np
import joblib
import os
import plotly.graph_objects as go
import plotly.express as px

# 1. Konfigurasi Tampilan Halaman Web
st.set_page_config(
    page_title="Titanic Survival Predictor AI",
    page_icon="🚢",
    layout="wide",
    initial_sidebar_state="expanded"
)

# 2. Fungsi Pemuatan Model dengan Caching
@st.cache_resource
def muat_model_pipeline():
    nama_file = "model_prediksi_titanic_pipeline.joblib"
    if os.path.exists(nama_file):
        return joblib.load(nama_file)
    else:
        st.error(f"⚠️ Berkas model '{nama_file}' tidak ditemukan di server!")
        return None

pipeline_model = muat_model_pipeline()

# 3. Header & Deskripsi Aplikasi
st.title("🚢 Titanic Survival Predictor AI")
st.markdown("Aplikasi web ini mendemonstrasikan integrasi **Fullstack Data Mining**: Menghubungkan model *Machine Learning Pipeline* (Scikit-Learn) ke dalam antarmuka web interaktif berbasis *Streamlit* dan visualisasi dinamis *Plotly*.")

st.write("---")

# 4. Panel Navigasi Sidebar (Masukan Profil Penumpang)
st.sidebar.image("https://raw.githubusercontent.com/tandpfun/skill-icons/main/icons/Python-Dark.svg", width=50)
st.sidebar.header("📋 Profil Calon Penumpang")

p_sex = st.sidebar.selectbox("Jenis Kelamin:", ["wanita", "pria"], help="Pilih jenis kelamin penumpang")
p_age = st.sidebar.slider("Usia Penumpang (Tahun):", min_value=1, max_value=80, value=25, help="Geser untuk menentukan usia")
p_class = st.sidebar.selectbox("Kelas Kabin Tiket:", ["VIP", "Bisnis", "Ekonomi"], help="Kelas tiket perjalanan kapal")

# Saran tarif default yang realistis berdasarkan kelas kabin
tarif_default = 90.0 if p_class == "VIP" else (30.0 if p_class == "Bisnis" else 10.0)
p_fare = st.sidebar.number_input("Tarif Tiket ($ USD):", min_value=0.0, max_value=600.0, value=float(tarif_default), step=5.0)

st.sidebar.markdown("---")
st.sidebar.caption("💡 Dikembangkan dengan Python & Streamlit • Data Mining Zero to Hero")

# 5. Tab Menu Aplikasi
tab_single, tab_batch, tab_about = st.tabs([
    "🎯 Prediksi Individu", 
    "📂 Prediksi Massal (Batch CSV)", 
    "ℹ️ Arsitektur AI & Pipeline"
])

# ==================== TAB 1: PREDIKSI INDIVIDU ====================
with tab_single:
    col_input, col_result = st.columns([1, 1], gap="medium")
    
    with col_input:
        st.subheader("📋 Ringkasan Parameter Masukan")
        df_display = pd.DataFrame([{
            "Parameter": "Jenis Kelamin", "Nilai": p_sex.capitalize()
        }, {
            "Parameter": "Usia", "Nilai": f"{p_age} Tahun"
        }, {
            "Parameter": "Kelas Kabin", "Nilai": f"Kelas {p_class}"
        }, {
            "Parameter": "Tarif Tiket", "Nilai": f"${p_fare:,.2f} USD"
        }])
        st.table(df_display)
        
        tombol_hitung = st.button("🔮 Hitung Estimasi Keselamatan AI", type="primary", use_container_width=True)

    with col_result:
        st.subheader("📊 Hasil Analisis Kecerdasan Buatan")
        
        if tombol_hitung:
            if pipeline_model is not None:
                # Siapkan data mentah berformat DataFrame yang cocok dengan Pipeline Modul 09
                data_penumpang = pd.DataFrame([{
                    'Umur': float(p_age),
                    'Tarif_Tiket': float(p_fare),
                    'Jenis_Kelamin': p_sex,
                    'Kelas_Kabin': p_class
                }])
                
                # Eksekusi inferensi nyata melalui pipeline model
                prediksi_kelas = pipeline_model.predict(data_penumpang)[0]
                prob_selamat = pipeline_model.predict_proba(data_penumpang)[0][1]
                prob_persen = prob_selamat * 100
                
                if prediksi_kelas == 1:
                    st.success("### 🎉 STATUS: DIPREDIKSI SELAMAT")
                    st.markdown(f"**Peluang Keselamatan:** `{prob_persen:.1f}%` (Tinggi)")
                else:
                    st.error("### ⚠️ STATUS: RISIKO TINGGI (TIDAK SELAMAT)")
                    st.markdown(f"**Peluang Keselamatan:** `{prob_persen:.1f}%` (Rendah)")
                
                # Visualisasi Interaktif Plotly: Speedometer Gauge Chart
                warna_bar = "#2ca02c" if prob_selamat >= 0.5 else "#d62728"
                fig_gauge = go.Figure(go.Indicator(
                    mode="gauge+number",
                    value=prob_persen,
                    number={'suffix': "%", 'font': {'size': 32}},
                    domain={'x': [0, 1], 'y': [0, 1]},
                    title={'text': "Indikator Peluang Selamat", 'font': {'size': 18}},
                    gauge={
                        'axis': {'range': [0, 100], 'tickwidth': 1, 'tickcolor': "gray"},
                        'bar': {'color': warna_bar},
                        'steps': [
                            {'range': [0, 50], 'color': "rgba(214, 39, 40, 0.15)"},
                            {'range': [50, 100], 'color': "rgba(44, 160, 44, 0.15)"}
                        ],
                        'threshold': {
                            'line': {'color': "black", 'width': 3},
                            'thickness': 0.75,
                            'value': 50
                        }
                    }
                ))
                fig_gauge.update_layout(height=260, margin=dict(l=20, r=20, t=30, b=20))
                st.plotly_chart(fig_gauge, use_container_width=True)
                
                # Penjelasan Kontekstual Ramah Awam
                with st.expander("💡 Mengapa Model Memberikan Prediksi Ini?"):
                    if p_sex == "wanita":
                        st.write("• **Protokol Maritim**: Penumpang wanita mendapat prioritas utama pengisian sekoci penyelamat (*Women and children first*).")
                    else:
                        st.write("• **Keterbatasan Sekoci**: Penumpang pria dewasa mengalami tingkat kesulitan evakuasi lebih tinggi karena keterbatasan kuota sekoci.")
                    
                    if p_class == "VIP":
                        st.write("• **Posisi Kabin**: Dek kabin VIP berada di lantai atas kapal, memberikan akses evakuasi tercepat menuju geladak utama.")
                    elif p_class == "Ekonomi":
                        st.write("• **Lokasi Dek Bawah**: Kabin kelas ekonomi terletak di bagian lambung bawah kapal dengan lorong evakuasi yang lebih rumit saat air mulai masuk.")
            else:
                st.warning("Model belum termuat dengan sempurna.")
        else:
            st.info("👈 Silakan pilih profil penumpang di sebelah kiri, lalu klik tombol **'Hitung Estimasi Keselamatan AI'**.")

# ==================== TAB 2: PREDIKSI MASSAL (BATCH) ====================
with tab_batch:
    st.subheader("📂 Prediksi Sekaligus Banyak Penumpang via File CSV")
    st.markdown("Fitur ini dirancang untuk memproses ratusan data penumpang secara serentak. Unggah file CSV yang memiliki kolom: `Umur`, `Tarif_Tiket`, `Jenis_Kelamin`, `Kelas_Kabin`.")
    
    # Tombol Unduh File Sampel
    if os.path.exists("penumpang_uji_batch.csv"):
        with open("penumpang_uji_batch.csv", "rb") as f:
            st.download_button(
                label="📥 Unduh Contoh File CSV Uji Coba (10 Penumpang)",
                data=f,
                file_name="contoh_penumpang_batch.csv",
                mime="text/csv",
                help="Klik untuk mengunduh template CSV siap uji"
            )
    
    file_unggahan = st.file_uploader("Unggah file CSV Anda ke sini:", type=["csv"])
    
    if file_unggahan is not None:
        try:
            df_batch = pd.read_csv(file_unggahan)
            st.write("📋 **Pratinjau Data Unggahan:**")
            st.dataframe(df_batch.head(5), use_container_width=True)
            
            kolom_wajib = ['Umur', 'Tarif_Tiket', 'Jenis_Kelamin', 'Kelas_Kabin']
            kolom_cocok = all(col in df_batch.columns for col in kolom_wajib)
            
            if kolom_cocok and pipeline_model is not None:
                # Eksekusi prediksi massal
                X_batch = df_batch[kolom_wajib]
                batch_preds = pipeline_model.predict(X_batch)
                batch_probas = pipeline_model.predict_proba(X_batch)[:, 1]
                
                df_hasil = df_batch.copy()
                df_hasil['Prediksi_AI'] = np.where(batch_preds == 1, 'Selamat 🎉', 'Tidak Selamat ⚠️')
                df_hasil['Peluang_Selamat (%)'] = (batch_probas * 100).round(1)
                
                st.success(f"✅ Berhasil memproses seluruh {len(df_hasil)} baris data calon penumpang!")
                
                col_chart, col_stat = st.columns([1, 1])
                
                with col_chart:
                    # Visualisasi Plotly Donut Chart
                    status_counts = pd.Series(np.where(batch_preds == 1, 'Selamat', 'Tidak Selamat')).value_counts()
                    fig_donut = px.pie(
                        values=status_counts.values,
                        names=status_counts.index,
                        title="Proporsi Hasil Prediksi Evakuasi",
                        hole=0.45,
                        color=status_counts.index,
                        color_discrete_map={'Selamat': '#2ca02c', 'Tidak Selamat': '#d62728'}
                    )
                    fig_donut.update_layout(height=280, margin=dict(l=20, r=20, t=40, b=20))
                    st.plotly_chart(fig_donut, use_container_width=True)
                
                with col_stat:
                    total_selamat = int((batch_preds == 1).sum())
                    total_korban = int((batch_preds == 0).sum())
                    st.metric("Total Penumpang Diprediksi Selamat", f"{total_selamat} Orang", f"{(total_selamat/len(df_hasil))*100:.1f}%")
                    st.metric("Total Penumpang Berisiko Tinggi", f"{total_korban} Orang", f"{(total_korban/len(df_hasil))*100:.1f}%", delta_color="inverse")
                
                st.write("📊 **Tabel Hasil Prediksi Lengkap:**")
                st.dataframe(df_hasil, use_container_width=True)
                
                csv_hasil = df_hasil.to_csv(index=False).encode('utf-8')
                st.download_button(
                    label="💾 Unduh Hasil Prediksi Lengkap (CSV)",
                    data=csv_hasil,
                    file_name="hasil_analisis_titanic_ai.csv",
                    mime="text/csv",
                    type="primary"
                )
            else:
                st.error(f"⚠️ Berkas CSV harus memuat kolom: {kolom_wajib}")
        except Exception as e:
            st.error(f"Terjadi kesalahan saat memproses file CSV: {e}")

# ==================== TAB 3: TENTANG ARSITEKTUR ====================
with tab_about:
    st.subheader("🧠 Arsitektur Pipeline di Balik Layar")
    st.markdown("""
    Aplikasi ini dibangun menggunakan arsitektur **Zero Data Leakage** dari Scikit-Learn:
    
    1. **Prapemrosesan Numerik (`Umur`, `Tarif_Tiket`)**:
       * `SimpleImputer(strategy='median')`: Mengisi nilai kosong secara objektif.
       * `StandardScaler()`: Menstandarkan skala fitur ke rata-rata 0 dan variansi 1.
    2. **Prapemrosesan Kategorikal (`Jenis_Kelamin`, `Kelas_Kabin`)**:
       * `SimpleImputer(strategy='most_frequent')`: Mengisi kategori yang hilang.
       * `OneHotEncoder(handle_unknown='ignore')`: Mengubah teks menjadi representasi biner.
    3. **Mesin Klasifikasi (*Estimator*)**:
       * `RandomForestClassifier(n_estimators=100)`: Menggabungkan keputusan 100 pohon keputusan secara demokratis.
    
    ---
    📚 *Bagian dari Kurikulum Data Mining: From Zero to Hero oleh Anton Prafanto.*
    """)
````


## 🛠️ 8. Panduan Mengatasi Masalah Umum (*Troubleshooting FAQ*)

Saat menjalankan aplikasi web data science pertama kali, pemula sering menemui beberapa kendala teknis jaringan. Jangan panik! Berikut adalah solusinya:

| Gejala Masalah | Akar Penyebab | Langkah Solusi Cepat |
| :--- | :--- | :--- |
| **`Port 8501 is already in use`** | Server Streamlit sebelumnya masih berjalan di background Colab. | Jalankan baris perintah ini di sel baru Colab untuk mematikan proses lama: `!pkill -f streamlit` lalu jalankan ulang. |
| **Halaman Localtunnel meminta IP tapi ditolak** | Sesi Colab Anda baru saja terputus atau restart sehingga IP publiknya berganti baru. | Jalankan ulang sel kode pendeteksi IP publik di atas untuk menyalin angka IP teranyar. |
| **Localtunnel macet / `504 Gateway Timeout`** | Server proxy Localtunnel internasional sedang mengalami antrean padat. | Gunakan opsi kedua: **Ngrok** (`pyngrok`) atau gunakan argumen subdomain kustom: `!npx localtunnel --port 8501 --subdomain web-titanic-saya`. |
| **Error `ModuleNotFoundError` saat Deploy ke Cloud** | Ada pustaka yang Anda pakai di kode Python tapi lupa dituliskan di `requirements.txt`. | Buka repositori GitHub Anda, edit file `requirements.txt`, tambahkan pustaka yang hilang (contoh: `plotly>=5.15.0`), lalu simpan (*commit*).
