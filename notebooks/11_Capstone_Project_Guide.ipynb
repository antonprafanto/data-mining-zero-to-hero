## 📋 1. Checklist Eksekusi Alur CRISP-DM Lengkap

Ikuti 6 langkah ini di dalam notebook proyek akhir Anda:

### Step 1: Business Understanding
- Apa latar belakang masalah yang ingin diselesaikan?
- Apa tujuan utama model data mining ini bagi pengguna/perusahaan?

### Step 2: Data Understanding
- Dari mana dataset diperoleh (Kaggle/UCI ML)?
- Berapa jumlah baris dan kolom? Apa saja arti setiap kolom?
- Eksplorasi Data Awal (EDA) dan visualisasi grafik sebaran target.

### Step 3: Data Preparation
- Penanganan nilai hilang (*missing values*).
- Deteksi & penanganan pencilan (*outliers*).
- Penskalaan fitur (*StandardScaler / MinMaxScaler*).
- Pengkodean variabel kategori (*One-Hot / Ordinal Encoding*).

### Step 4: Modeling
- Membagi data latih dan data uji (`train_test_split`).
- Melatih minimal **2 atau 3 algoritma** yang berbeda (contoh: Random Forest vs Logistic Regression vs KNN).
- Melakukan *Hyperparameter Tuning* dengan `GridSearchCV`.

### Step 5: Evaluation
- Membandingkan performa model dengan metrik yang tepat (Accuracy, F1-Score, RMSE).
- Menganalisis *Confusion Matrix* dan kesalahan prediksi.
- Menyimpan model pipeline terbaik ke file `.joblib`.

### Step 6: Deployment
- Membuat antarmuka pengguna interaktif dengan **Streamlit**.
- Mempublikasikan aplikasi ke **Streamlit Community Cloud**.
