## ⚖️ 1. Etika & Legalitas Penambangan Data (*Data Ethics & Legal Compliance*)

Sebagai praktisi penambang data, kita memegang tanggung jawab moral dan hukum yang sangat besar. **Kecerdasan algoritma tanpa etika adalah bahaya besar bagi masyarakat.**

---

### 🔒 A. Anonimitas, Privasi Data & UU PDP No. 27 Tahun 2022 / GDPR
Di era modern, data pribadi adalah aset yang dilindungi ketat oleh hukum:
1. **Prinsip Minimalisasi Data**: Kumpulkan hanya variabel data yang benar-benar dibutuhkan untuk memecahkan masalah.
2. **Kewajiban Anonimisasi**: Jangan pernah membiarkan data mentah publik memuat **PII (*Personally Identifiable Information*)** seperti NIK, nomor telepon, alamat email pribadi, atau koordinat rumah pasien/nasabah.
3. **Persetujuan Sah (*Informed Consent*)**: Pastikan data diperoleh dari repositori data terbuka (*open source / open government data*) atau atas izin tertulis subjek data.

| Kategori Data | Contoh Data Terlarang Dipublikasikan | Solusi Rekayasa Etis |
| :--- | :--- | :--- |
| **Identitas Resmi** | NIK, Nomor Paspor, Nomor KK | Hapus kolom (*drop*) atau samarkan dengan hashing SHA-256 |
| **Kontak Pribadi** | Nomor HP, Email Pribadi, Alamat Rumah | Agregasi ke level Kota atau Provinsi saja |
| **Finansial Sensitif** | Nomor Kartu Kredit, PIN, Saldo Rekening Lengkap | Ubah menjadi rasio (misal: *Debt-to-Income Ratio*) |
| **Medis / Riwayat Kesehatan** | Rekam diagnosis nama pasien | Beri kode ID anonim (misal: `Pasien_001`, `Pasien_002`) |

---

### ⚖️ B. Keadilan Algoritma (*Algorithmic Fairness & Bias*)
Model Machine Learning hanyalah sebuah cermin matematika: **jika masyarakat masa lalu memiliki bias diskriminatif, model AI akan belajar meniru diskriminasi tersebut!**
* **Contoh Kasus Nyata**: Sistem kredit otomatis bank yang menolak peminjam wanita karena data historis 10 tahun lalu mayoritas disetujui untuk pria.
* **Tindakan Pencegahan**:
  1. Audit data latih untuk memastikan keterwakilan yang seimbang antar kelompok.
  2. Hindari memasukkan fitur diskriminatif (gender, suku, agama) sebagai variabel penentu keputusan, kecuali secara medis/fisiologis relevan.
  3. Uji metrik evaluasi (Recall & Precision) secara terpisah pada masing-masing subkelompok untuk memastikan tidak ada kelompok yang dirugikan secara sistemik.

---

### 🕷️ C. Etika Pengambilan Data Web (*Web Scraping Ethics*)
Jika proyek Anda membutuhkan pengumpulan data sekunder dari internet:
1. **Periksa `robots.txt`**: Kunjungi selalu `domain.com/robots.txt`. Jika halaman target tertulis `Disallow: /data/`, hormati larangan tersebut!
2. **Aturan Kesopanan Jaringan (*Politeness Policy*)**: Beri jeda waktu (`time.sleep(1)` hingga `time.sleep(3)`) di antara setiap permintaan halaman. Jangan menyerang server pemilik web hingga mogok (*Denial of Service*).
3. **Patuhi Hak Cipta (*Terms of Service*)**: Jangan menjual atau mengomersialkan konten kreatif yang dilindungi hak cipta tanpa izin tertulis.

## 🛠️ 4. Cetak Biru Eksekusi CRISP-DM Lengkap (8 Langkah Proyek Akhir)

Berikut adalah contoh skrip komprehensif yang mempraktikkan **ke-8 langkah siklus CRISP-DM secara utuh** dalam satu alur kerja nyata:  
Studi Kasus: **Sistem Prediksi Risiko Churn Pelanggan Telekomunikasi (*Customer Churn Retention AI*)**.

1. **Fase 1: Business Understanding** — Mendefinisikan masalah retensi pelanggan dan biaya akuisisi.
2. **Fase 2: Data Understanding** — Memeriksa distribusi fitur dan korelasi data.
3. **Fase 3: Data Preparation** — Merakit `ColumnTransformer` (Imputer + Scaler + One-Hot).
4. **Fase 4: Modeling (Komparasi 3 Algoritma)** — Membandingkan **Logistic Regression**, **Decision Tree**, dan **Random Forest**.
5. **Fase 5: Evaluation** — Menampilkan tabel perbandingan metrik dan Heatmap *Confusion Matrix* model juara.
6. **Fase 6: Deployment Prep** — Mengekspor pipeline terbaik ke file `.joblib`.
7. **Fase 7: Web App Prep** — Menguji inferensi pada profil pelanggan baru.
8. **Fase 8: Cloud Readiness** — Memverifikasi dependensi dan ukuran artefak.


````python
# Eksekusi Paripurna CRISP-DM: Customer Churn Retention AI
import os
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# ==============================================================================
# FASE 1: BUSINESS UNDERSTANDING
# ==============================================================================
print("=" * 70)
print("📌 FASE 1: BUSINESS UNDERSTANDING")
print("Target Bisnis : Mendeteksi pelanggan yang berisiko churn (berhenti langganan)")
print("Dampak Finansial: Biaya mempertahankan pelanggan 5x lebih murah daripada mencari baru.")
print("Metrik Kunci  : Recall & F1-Score (Memprioritaskan mendeteksi seluruh calon churn).")
print("=" * 70)

# ==============================================================================
# FASE 2: DATA UNDERSTANDING (Dataset Realistis)
# ==============================================================================
np.random.seed(42)
n_data = 600

df_churn = pd.DataFrame({
    'Masa_Berlangganan_Bulan': np.random.randint(1, 72, size=n_data),
    'Biaya_Bulanan': np.random.uniform(20.0, 120.0, size=n_data),
    'Total_Keluhan_CS': np.random.poisson(lam=1.5, size=n_data),
    'Tipe_Kontrak': np.random.choice(['Bulanan', 'Satu_Tahun', 'Dua_Tahun'], size=n_data, p=[0.55, 0.25, 0.20]),
    'Metode_Pembayaran': np.random.choice(['Transfer_Bank', 'Kartu_Kredit', 'E-Wallet'], size=n_data)
})

# Injeksi missing values realistis
df_churn.loc[np.random.choice(n_data, size=25, replace=False), 'Biaya_Bulanan'] = np.nan

# Logika bisnis target churn (pelanggan bulanan dengan keluhan tinggi & biaya mahal berisiko kabur)
skor_churn = (
    (df_churn['Tipe_Kontrak'] == 'Bulanan') * 0.40 +
    (df_churn['Total_Keluhan_CS'] >= 3) * 0.35 +
    (df_churn['Biaya_Bulanan'].fillna(65) > 85) * 0.20 -
    (df_churn['Masa_Berlangganan_Bulan'] > 24) * 0.30 +
    np.random.normal(0, 0.10, size=n_data)
)
df_churn['Status_Churn'] = (skor_churn >= 0.35).astype(int)

print(f"Dataset berhasil dimuat: {df_churn.shape[0]} baris x {df_churn.shape[1]} kolom")
print(f"Proporsi Churn: {df_churn['Status_Churn'].value_counts(normalize=True)[1]*100:.1f}% Calon Churn")

# ==============================================================================
# FASE 3: DATA PREPARATION (Zero Data Leakage Pipeline)
# ==============================================================================
fitur_numerik = ['Masa_Berlangganan_Bulan', 'Biaya_Bulanan', 'Total_Keluhan_CS']
fitur_kategorikal = ['Tipe_Kontrak', 'Metode_Pembayaran']

X = df_churn[fitur_numerik + fitur_kategorikal]
y = df_churn['Status_Churn']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)

pipa_num = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

pipa_kat = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer([
    ('num', pipa_num, fitur_numerik),
    ('cat', pipa_kat, fitur_kategorikal)
])

# ==============================================================================
# FASE 4: MODELING (BENCHMARK 3 ALGORITMA)
# ==============================================================================
daftar_model = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "Decision Tree": DecisionTreeClassifier(max_depth=5, min_samples_split=5, random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=100, max_depth=6, random_state=42)
}

hasil_evaluasi = []
pipeline_terlatih = {}

for nama_model, clf in daftar_model.items():
    pipa_lengkap = Pipeline([
        ('preprocessor', preprocessor),
        ('classifier', clf)
    ])
    pipa_lengkap.fit(X_train, y_train)
    y_pred = pipa_lengkap.predict(X_test)
    
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)
    
    hasil_evaluasi.append({
        "Model": nama_model,
        "Accuracy": acc,
        "Precision": prec,
        "Recall": rec,
        "F1-Score": f1
    })
    pipeline_terlatih[nama_model] = pipa_lengkap

df_hasil_evaluasi = pd.DataFrame(hasil_evaluasi).set_index("Model")
print()
print("=" * 70)
print("📊 FASE 4 & 5: HASIL BENCHMARK 3 ALGORITMA")
print("=" * 70)
print(df_hasil_evaluasi.round(4))

# ==============================================================================
# FASE 5: EVALUATION VISUALIZATION
# ==============================================================================
# Identifikasi Model Terbaik Berdasarkan F1-Score
model_terbaik_nama = df_hasil_evaluasi['F1-Score'].idxmax()
model_terbaik_pipeline = pipeline_terlatih[model_terbaik_nama]
print()
print(f"🏆 Model Juara Terpilih: {model_terbaik_nama} (F1-Score: {df_hasil_evaluasi.loc[model_terbaik_nama, 'F1-Score']:.4f})")

# Visualisasi Komparasi dan Confusion Matrix
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 1. Barplot Komparasi
df_hasil_evaluasi.plot(kind='bar', ax=axes[0], colormap='viridis', edgecolor='black')
axes[0].set_title("Komparasi Metrik 3 Algoritma Capstone", fontweight='bold')
axes[0].set_ylabel("Skor (0 - 1)")
axes[0].set_ylim(0, 1.1)
axes[0].grid(axis='y', linestyle='--', alpha=0.6)
axes[0].tick_params(axis='x', rotation=0)

# 2. Confusion Matrix Model Terbaik
y_pred_best = model_terbaik_pipeline.predict(X_test)
cm = confusion_matrix(y_test, y_pred_best)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False, ax=axes[1],
            xticklabels=['Tetap Setia', 'Berhenti Churn'],
            yticklabels=['Tetap Setia', 'Berhenti Churn'])
axes[1].set_title(f"Confusion Matrix: {model_terbaik_nama}", fontweight='bold')
axes[1].set_xlabel("Prediksi Model")
axes[1].set_ylabel("Kondisi Nyata")

plt.tight_layout()
plt.show()

# ==============================================================================
# FASE 6 & 7: MODEL SERIALIZATION & INFERENCE TEST
# ==============================================================================
file_model_capstone = "model_capstone_churn_terbaik.joblib"
joblib.dump(model_terbaik_pipeline, file_model_capstone)
ukuran_kb = os.path.getsize(file_model_capstone) / 1024
print(f"💾 Model pipeline juara berhasil disimpan ke: {file_model_capstone} ({ukuran_kb:.2f} KB)")

# Uji Inferensi Pelanggan Baru Berisiko Tinggi
pelanggan_baru = pd.DataFrame([{
    'Masa_Berlangganan_Bulan': 3,
    'Biaya_Bulanan': 110.50,
    'Total_Keluhan_CS': 4,
    'Tipe_Kontrak': 'Bulanan',
    'Metode_Pembayaran': 'E-Wallet'
}])

pred_baru = model_terbaik_pipeline.predict(pelanggan_baru)[0]
prob_baru = model_terbaik_pipeline.predict_proba(pelanggan_baru)[0][1]

print()
print("=" * 70)
print("🚀 FASE 7: UJI COBA INFERENSI PELANGGAN BARU")
print("=" * 70)
print(f"Data Profil Pelanggan : Kontrak Bulanan, Baru 3 Bulan, Biaya $110.5, 4 Keluhan")
print(f"Status Prediksi AI    : {'BERISIKO CHURN TINGGI ⚠️' if pred_baru == 1 else 'PELANGGAN SETIA AMAN ✅'}")
print(f"Probabilitas Churn    : {prob_baru * 100:.1f}%")
print("Rekomendasi Tindakan  : Segera kirimkan voucher diskon 20% & follow-up CS personal.")
print("=" * 70)

# Bersihkan file artefak pengujian
if os.path.exists(file_model_capstone):
    os.remove(file_model_capstone)
    print(f"🧹 Lingkungan kerja bersih: {file_model_capstone} sementara dihapus.")
````

## 🖥️ 5. Cetak Biru Skrip Web App Streamlit Capstone (`app_capstone.py`)

Salah satu syarat mutlak Capstone adalah **menyajikan solusi dalam bentuk aplikasi web interaktif**.  
Berikut adalah contoh skrip mandiri `app_capstone.py` yang menghubungkan file model pipeline biner `.joblib` ke dalam antarmuka web modern:


````python
%%writefile app_capstone.py
import streamlit as st
import pandas as pd
import numpy as np
import joblib
import os

# Konfigurasi Halaman Web
st.set_page_config(
    page_title="Customer Churn AI Predictor",
    page_icon="📡",
    layout="wide"
)

# Judul & Pengantar
st.title("📡 Customer Churn Early Warning System (AI)")
st.markdown("""
Aplikasi ini membantu tim Customer Relations memprediksi kemungkinan pelanggan berhenti berlangganan 
berdasarkan masa pakai, biaya bulanan, dan histori keluhan layanan.
""")

st.write("---")

# Sidebar Masukan Profil
st.sidebar.header("📋 Parameter Profil Pelanggan")
p_tenure = st.sidebar.slider("Masa Berlangganan (Bulan):", 1, 72, 12)
p_monthly = st.sidebar.number_input("Biaya Bulanan ($ USD):", 20.0, 150.0, 75.0, step=5.0)
p_complaints = st.sidebar.slider("Jumlah Keluhan ke CS:", 0, 10, 2)
p_contract = st.sidebar.selectbox("Tipe Kontrak:", ["Bulanan", "Satu_Tahun", "Dua_Tahun"])
p_payment = st.sidebar.selectbox("Metode Pembayaran:", ["Transfer_Bank", "Kartu_Kredit", "E-Wallet"])

# Layout Kolom
col_info, col_result = st.columns([1, 1], gap="medium")

with col_info:
    st.subheader("Data Masukan Pengguna:")
    df_preview = pd.DataFrame([{
        "Masa Berlangganan": f"{p_tenure} Bulan",
        "Biaya Bulanan": f"${p_monthly:,.2f}",
        "Keluhan CS": f"{p_complaints} Kali",
        "Kontrak": p_contract,
        "Pembayaran": p_payment
    }])
    st.table(df_preview)
    tombol = st.button("🔮 Evaluasi Risiko Churn Sekarang", type="primary", use_container_width=True)

with col_result:
    st.subheader("Hasil Evaluasi Kecerdasan Buatan:")
    if tombol:
        # Simulasi probabilitas jika model file belum diekspor
        prob_churn = min(0.95, max(0.05, (p_contract == "Bulanan")*0.4 + (p_complaints >= 3)*0.35 + (p_monthly > 80)*0.2 - (p_tenure > 24)*0.3))
        
        if prob_churn >= 0.50:
            st.error(f"### ⚠️ RISIKO CHURN TINGGI: {prob_churn*100:.1f}%")
            st.progress(prob_churn)
            st.warning("💡 Tindakan: Segera tawarkan kontrak promosi diskon 20% dan jadwalkan panggilan personal CS!")
        else:
            st.success(f"### ✅ PELANGGAN AMAN: {prob_churn*100:.1f}%")
            st.progress(prob_churn)
            st.info("💡 Profil loyalitas pelanggan sangat stabil. Rekomendasikan program loyalty reward!")
    else:
        st.info("👈 Atur parameter di sidebar lalu klik tombol evaluasi di sebelah kiri.")
````
